# Price Regression with PyCaret
This notebook uses PyCaret to build a regression model for predicting prices based on various features from the `real_estate.csv` dataset.

**Dataset:**
- `real_estate.csv`: Contains various features of real estate property and their prices.

**Steps:**
1. Load the dataset
2. Set up PyCaret regression experiment
3. Compare models and select the best
4. Evaluate and save the best model
5. Predict prices of real estate properties

```python
# Import required libraries
import pandas as pd
import numpy as np
```

```python
# Load the dataset
data = pd.read_csv(r'D:\Pycaret(1)\pycaret_assignment\regression\real_estate_dataset.csv')
data.head()
```

```python
# PyCaret Regression Setup
from pycaret.regression import RegressionExperiment
r = RegressionExperiment()
r.setup(data=data, target='Price', session_id=123, normalize=True, transformation=True, use_gpu=True)
```

```python
# Compare models and select the best
best_model = r.compare_models()
```

```python
# Evaluate the best model
r.evaluate_model(best_model)
```

```python
# Save the best model for future use
r.save_model(best_model, 'best_price_model')
```

```python
# Predict car prices on the same data (or new/unseen data)
predictions = r.predict_model(best_model)
predictions.head()
```
