# Anomaly Detection with PyCaret
This notebook uses PyCaret to perform unsupervised anomaly detection on `financial_fraud_detection_dataset.csv`.

**Dataset:**
- Transactional and customer features for 2,512 samples.

**Steps:**
1. Load the dataset
2. Set up PyCaret anomaly detection experiment
3. Create and evaluate an anomaly detection model
4. Assign anomaly labels and save the model

```python
# Import required libraries
import pandas as pd
import numpy as np
```

```python
# Load the dataset
data = pd.read_csv(r'D:\Pycaret(1)\pycaret_assignment\anomaly_detection\bank_transactions_data_2.csv')
data.head()
```

```python
# PyCaret anomaly detection setup
from pycaret.anomaly import AnomalyExperiment
a = AnomalyExperiment()
a.setup(data=data, session_id=123, normalize=True, use_gpu=True)
```

```python
# Create an anomaly detection model (Isolation Forest)
knn = a.create_model('knn')
# Evaluate the model (interactive, notebook only)
a.evaluate_model(knn)
```

```python
# Assign anomaly labels to the data
anomaly_results = a.assign_model(knn)
anomaly_results.head()
```

```python
a.plot_model(knn,plot='umap')
```

```python
# Save the model for future use
a.save_model(knn, 'best_anomaly_model')
```
