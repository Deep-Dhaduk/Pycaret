```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv('D:\Pycaret(1)\pycaret_assignment\multiclass\Mobile_price_data.csv')
data.head()
```

```python
from pycaret.classification import ClassificationExperiment
p = ClassificationExperiment()
```

```python
p.setup(data=data,
        target='price_range',
        session_id=123,
        normalize=True,
        transformation=True,
        use_gpu=True)
```

```python
best_model = p.compare_models()
```

```python
p.evaluate_model(best_model)
```

```python
# Save the best model for future use
p.save_model(best_model, 'best_student_performance_model')
```

```python
# Predict on the same data (or you can use new/unseen data)
predictions = p.predict_model(best_model)
predictions.head()
```
